## What is ASN.1?
Abstract Syntax Notation One (ASN.1) is a standardized language used to describe data structures. Its primary purpose is to enable the exchange of complex data between different computer systems, regardless of their hardware, software, or programming languages.
ASN.1 pove kakšna je struktura podatkov, encoding rules pa povedo kako se ta struktura pretvori v bajte.

## Basic ASN.1 Data Types
ASN.1 provides a set of built-in fundamental data types that serve as the building blocks for more complex structures. Some common basic types include:

BOOLEAN: Represents a logical value (TRUE or FALSE).

INTEGER: Represents whole numbers.

REAL: Represents real (floating-point) numbers.

BIT STRING: Represents a sequence of bits.

OCTET STRING: Represents a sequence of bytes.

NULL: Represents the absence of a value.

ENUMERATED: Defines a type with a limited set of named integer values.

Character String Types: Various types for text, such as UTF8String, IA5String, PrintableString, etc.

Time Types: Types for date and time, like UTCTime and GeneralizedTime.

## Structured Data Types and Example
ASN.1 also provides constructors to build complex structures. A fundamental one is SEQUENCE, which defines an ordered collection of elements, potentially of different types.

## Abstract vs. Transfer Syntax
ASN.1 separates the abstract syntax (the definition of the data structure itself, like the PersonIdentifier example above) from the transfer syntax (the rules for converting that structure into a sequence of bytes for transmission or storage).

Encoding: The process of converting a data structure (defined by the abstract syntax) into a binary or text representation using a specific transfer syntax (like BER (General purpose streaming), DER(Security, Certificates, Crypto), PER(Bandwidth-heavy (5G, Satellite)), XER(Legacy web systems), or JER(Modern Web APIs)).

Decoding: The reverse process of converting the encoded binary or text data back into the original data structure based on the abstract syntax definition.

Why Matching Definitions is Crucial
For successful communication, both the sender and the receiver must use the exact same ASN.1 abstract syntax definition for the data being exchanged. The decoder relies on this definition to correctly interpret the incoming sequence of bytes according to the agreed-upon transfer syntax. If the definitions don't match, the decoding process will fail, as the structure, types, or order of elements will not be as expected.

```python
# measurement_data.py (Example file name)

from pyasn1.type import univ, namedtype

# Define the ASN.1 type for a list of REAL numbers
ListOfReals = univ.SequenceOf(componentType=univ.Real())

# Define the ASN.1 type for the MeasurementData SEQUENCE
class MeasurementData(univ.Sequence):
    componentType = namedtype.NamedTypes(
        # Define the 'firstArray' component, which is a ListOfReals
        namedtype.NamedType('firstArray', ListOfReals),

        # Define the 'secondArray' component, which is also a ListOfReals
        namedtype.NamedType('secondArray', ListOfReals)
    )
    
stvar = MeasurementData()

# Initialize the sequence and populate it
reals = univ.SequenceOf(componentType=univ.Real())
reals.extend([1.0, 2.0, 3.0]) # Use floats to be safe

stvar['firstArray'] = reals

print(stvar.prettyPrint())
```

```python
stvar['firstArray'] = ListOfReals.setComponents(1.0*2, 2.0*2, 3.0*2)
print(stvar.prettyPrint())
```

```python
stvar['firstArray'][:] = [5.0, 6.0, 7.0]
print(stvar.prettyPrint())
```
