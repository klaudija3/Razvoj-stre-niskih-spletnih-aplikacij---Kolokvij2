```python
import sys
import socketserver
from pyasn1.type import univ, namedval, constraint
from pyasn1.type import  namedtype, tag, char
from pyasn1.type import useful
from pyasn1.codec.ber import encoder, decoder


from signature import Measurement
print('Start')

class MyTCPHandler(socketserver.BaseRequestHandler):
#class MyTCPHandler(object):
    """
    The RequestHandler class for our server.

    It is instantiated once per connection to the server, and must
    override the handle() method to implement communication to the
    client.
    """
    def handle(self):
        if len(self.client_address) > 0:
            print('Client IP=%s' % self.client_address[0])
        self.request.settimeout(5)
 
        curr_recv = self.request.recv(1024000)
 
        try:
            print('Datalen=%d' % len(curr_recv))
            #CRUCIAL- here we check if the format is correct!!!
            dec, byte = decoder.decode(curr_recv,asn1Spec=Measurement())
            print(str(dec))
            print('ASN1 OK')
            ID = dec.getComponentByName('SensorID')
            print(f'ID={ID}')
            self.request.sendall('ASN1 format OK'.encode())
        except:
            print('ASN1 not ok')
            self.request.sendall('ASN1 format NOT OK'.encode())
        
        print('DONE! sent')
        
if __name__ == "__main__":
    HOST, PORT = "0.0.0.0", 22222
    server = socketserver.TCPServer((HOST, PORT), MyTCPHandler)
    server.serve_forever()
```
