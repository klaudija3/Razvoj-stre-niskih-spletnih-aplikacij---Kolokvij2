## b)

```python
# univ je modul iz knjižnice pyasn1, ki vsebuje osnovne ASN.1 tipe.

# Sequence je ASN.1 sestavljen tip, podoben struct v C ali class/dataclass v Pythonu.
# Vsebuje več poimenovanih polj v določenem vrstnem redu.

# to pomeni:
# Datum ::= SEQUENCE {
#     leto   INTEGER,
#     mesec  INTEGER,
#     dan    INTEGER
# }

# univ vsebuje osnovne ASN.1 tipe:

# univ.Integer()      # celo število
# univ.Boolean()      # true/false
# univ.Sequence()     # struktura z več polji
# univ.OctetString()  # bajti

# char vsebuje znakovne / tekstovne ASN.1 tipe:

# char.UTF8String()       # UTF-8 tekst
# char.PrintableString()  # omejen nabor znakov
# char.IA5String()        # ASCII tekst
class Datum(univ.Sequence):
    componentType = namedtype.NamedTypes(namedtype.NamedType('leto', univ.Integer()), namedtype.NamedType('mesec', univ.Integer()), namedtype.NamedType('dan', univ.Integer()))
```

```python
date = Datum()
```

```python
date.setComponentByName('leto', 2023)
date.setComponentByName('mesec', 11)
date.setComponentByName('dan', 20)
```

```python
date['leto'].prettyPrint(), date['mesec'].prettyPrint(), date['dan'].prettyPrint()
```

```python
class Dan(univ.Integer):
    subtypeSpec = univ.Integer.subtypeSpec + constraint.ValueRangeConstraint(1,31)
    
class Mesec(univ.Integer):
    subtypeSpec = univ.Integer.subtypeSpec + constraint.ValueRangeConstraint(1,12)
    
class Leto(univ.Integer):
    subtypeSpec = univ.Integer.subtypeSpec + constraint.ValueRangeConstraint(0,2025)
```

```python
class Datum2(univ.Sequence):
    componentType = namedtype.NamedTypes(namedtype.NamedType('leto', Leto()),
                                         namedtype.NamedType('mesec', Mesec()),
                                         namedtype.NamedType('dan', Dan()))
primer = Datum2()
primer['dan'] = 31
```
