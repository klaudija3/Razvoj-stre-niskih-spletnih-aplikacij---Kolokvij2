### What is gRPC?

gRPC, which stands for **gRPC Remote Procedure Calls**, is a high-performance, open-source framework developed by Google. It is used to enable communication between services in a distributed system. Here's what it offers:

1. **Cross-Platform Compatibility**: gRPC can be used across various programming languages and platforms.
2. **Efficient Communication**: Uses HTTP/2 for transport, which provides features like multiplexing, flow control, header compression, and low latency.
3. **Language-Agnostic**: Uses Protocol Buffers (protobuf) as its Interface Definition Language (IDL) and serialization format, making it easy to define service interfaces in a language-agnostic way.
4. **Bi-directional Streaming**: Supports bi-directional streaming, allowing for real-time communication between client and server.

### How gRPC Works

1. **Define a Service**: You start by defining a service in a `.proto` file, which includes the methods that can be called remotely along with their input and output message types.
2. **Generate Code**: You use a tool called `protoc` (Protocol Buffers compiler) to generate client and server code from the `.proto` file.
3. **Implement Server**: Implement the server by defining the behavior of the service methods.
4. **Create Client**: Implement the client to call the remote methods on the server.

### What is Proto3?

**Protocol Buffers (protobuf)** is a language-neutral, platform-neutral, extensible mechanism for serializing structured data. The `.proto` file is where you define the structure of your data and the services.

**proto3** is the third version of Protocol Buffers, offering several improvements and simplifications over previous versions. Here are some key aspects:

1. **Syntax Simplification**: The proto3 syntax is simpler and more concise compared to proto2.
2. **Defaults and Enums**: Fields have default values automatically, and enums always have a default value of the first item.
3. **No Required Fields**: In proto3, all fields are optional, unlike proto2, which had `required` and `optional` fields.
4. **Support for JSON**: Proto3 supports automatic conversion to and from JSON, which is useful for web APIs.

### How Proto3 is Used with gRPC

1. **Service Definition**: In the `.proto` file, you define the service with its methods and the messages they exchange. For example, you might define a `Greeter` service with a `SayHello` method that takes a `HelloRequest` and returns a `HelloReply`.
2. **Data Serialization**: Proto3 handles the serialization and deserialization of the structured data (messages) being exchanged between the client and server.
3. **Code Generation**: The `protoc` compiler generates code for both the client and server based on the service definition in the `.proto` file.

### Workflow Summary

1. **Define Your Service**: Write a `.proto` file that defines your service methods and message types.
2. **Generate Code**: Use `protoc` to generate Python code from your `.proto` file.
3. **Implement Your Server**: Write the server-side code to implement the service methods.
4. **Implement Your Client**: Write the client-side code to call the service methods on the server.

### Benefits of gRPC

- **Efficiency**: Uses Protocol Buffers for efficient serialization.
- **Scalability**: Built on HTTP/2, providing high-performance communication.
- **Ease of Use**: Simplifies API development with clearly defined services and messages.
- **Cross-Language Support**: Supports multiple programming languages, making it suitable for heterogeneous environments.

In essence, gRPC combined with proto3 provides a powerful framework for building robust, high-performance, and language-agnostic APIs for distributed systems.

### Language-agnostic framework
gRPC itself is a language-agnostic framework for remote procedure calls. It supports multiple programming languages, including Python, Java, Go, C++, and many others. This means you can write a gRPC server in one language and a client in another, and they can communicate seamlessly.

Protocol Buffers (protobuf) is also language-agnostic. The .proto files used to define message structures and services are the same regardless of the language used for implementation. The protoc compiler can generate code for various languages from the same .proto file, enabling consistent data structures and communication protocols across different parts of a system.

### Comparison between ASN.1 and Proto3:

### Overview
- **ASN.1**: Standard for defining data structures used in telecommunications and networking (e.g., X.509 certificates).
- **Proto3 (Protocol Buffers)**: Google's language-agnostic data serialization format used in RPC systems, particularly with gRPC.

### Key Differences

1. **Origins and Usage**
   - **ASN.1**: Telecommunications and network protocols.
   - **Proto3**: Microservices and cloud applications.

2. **Syntax**
   - **ASN.1**: Complex and feature-rich.
   - **Proto3**: Simple and easy to read.
   
   Example:
   - **ASN.1**: 
     ```asn.1
     Person ::= SEQUENCE {
         name    UTF8String,
         age     INTEGER
     }
     ```
   - **Proto3**:
     ```proto
     syntax = "proto3";

     message Person {
         string name = 1;
         int32 age = 2;
     }
     ```

3. **Encoding Rules**
   - **ASN.1**: Multiple encoding rules (BER, DER, PER).
   - **Proto3**: Single efficient binary format, also supports JSON.

4. **Tooling and Ecosystem**
   - **ASN.1**: Mature but complex tooling.
   - **Proto3**: Robust ecosystem, especially with gRPC.

5. **Complexity and Learning Curve**
   - **ASN.1**: More complex and harder to learn.
   - **Proto3**: Simpler and easier to learn.

6. **Data Model**
   - **ASN.1**: Rich and complex (SEQUENCE, CHOICE, SET).
   - **Proto3**: Simplified data structures (message, enum, repeated fields).

7. **Compatibility**
   - **ASN.1**: Complex backward and forward compatibility.
   - **Proto3**: Designed for easy backward and forward compatibility.

### Use Cases

- **ASN.1**: Best for telecommunications, network protocols, and secure communications.
- **Proto3**: Ideal for microservices, RPC systems, and cloud-based applications.

### Summary

- **ASN.1**: Powerful and flexible, but complex.
- **Proto3**: Simplified and performance-oriented, especially useful with gRPC.

Choosing between them depends on your project's complexity and specific needs.

- **Generated Code:** When you compile the `.proto` file using a Protocol Buffer compiler, it generates two Python files: `test_pb2.py` (usually named after your `.proto` file) and `test_pb2_grpc.py`.

- **`pb2_grpc` File:** This file (`test_pb2_grpc.py`) deals with gRPC service definitions. It contains the service classes (likely `Naloga1Servicer` in this case) and their associated RPC methods (like `Komunikacija`). These service classes and methods handle the communication logic between the client and server.
