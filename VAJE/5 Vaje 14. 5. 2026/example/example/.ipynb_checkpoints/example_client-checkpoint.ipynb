```python
import grpc
import example_pb2_grpc as pb2_grpc
import example_pb2 as pb2

channel = grpc.insecure_channel(f'127.0.0.1:10024')
stub = pb2_grpc.ExampleStub(channel)
```

gRPC je način, da se dva programa pogovarjata med sabo, podobno kot REST API, ampak bolj “strojno” in učinkovito.

Po domače:

REST API: klient pošlje HTTP zahtevek na URL
gRPC: klient pokliče funkcijo na drugem računalniku, kot da je lokalna

A gRPC stub acts as a client-side proxy for a server's gRPC service. It provides type-safe methods corresponding to the defined RPCs, abstracting low-level communication and enforcing message structures for a streamlined and robust client experience. Stubs are typically named `<service_name>Stub`.


```python
asd = pb2.Sporocilo()
asd.text = 'zdravoo'
response = stub.Komunikacija(asd)
print(response.text)
```

```python
field_name = response.DESCRIPTOR.fields_by_number[2].name
print(f"Field name for number 2: {field_name}")
```
