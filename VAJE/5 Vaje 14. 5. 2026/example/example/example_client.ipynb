```python
import grpc
import example_pb2_grpc as pb2_grpc
import example_pb2 as pb2

channel = grpc.insecure_channel(f'127.0.0.1:10025')
stub = pb2_grpc.ExampleStub(channel)
```

A gRPC stub acts as a client-side proxy for a server's gRPC service. It provides type-safe methods corresponding to the defined RPCs, abstracting low-level communication and enforcing message structures for a streamlined and robust client experience. Stubs are typically named `<service_name>Stub`.


```python
text = pb2.Text()
text.val  = 'zdravo'
```

```python
integer = pb2.Int()
integer.val  = 32
```

```python
floating = pb2.Floating()
floating.val  = 32.3
```

```python
empty = pb2.Empty()
print(empty)
```

```python
field_name = floating.DESCRIPTOR.fields_by_number[1].name
print(f"Field name: {field_name}")
```

```python
response = stub.SendRandomText(pb2.Empty())
print(response)
```

```python
for i in  stub.StreamRandomInt(pb2.Empty()):
    print(i)
```

```python
def iterator(array):
    for i in array:
        yield i
        
stub.ObtainRandomFloatStream(iterator([pb2.Floating(val=10.0),pb2.Floating(val=8.0),pb2.Floating(val=9.0)]))
```
