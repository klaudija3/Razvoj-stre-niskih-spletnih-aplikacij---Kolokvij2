Naredite svoj novi database v InfluxDB. Pazite, da bo imel unikatno ime.
Napišite Python program, ki vsake pol sekunde vanj pošlje tri točke.
• Prva točka je meritev hitrosti s fieldom ”h”
• Druga točka je meritev gostote s fieldom ”g”
• Tretja točka je meritev pritiska s fieldom ”p”
Vsaka točka naj v tagu vsebuje informacijo, na kateri izmed treh lokacij. Možne lokacije
so ”1”, ”2” in ”3”

```python
from influxdb import InfluxDBClient
import random
import time

client = InfluxDBClient(
    host='149.62.71.186',
    port=8086,
    username='admin',
    password='fisinflux'
)

db_name = "moje_ime_db"

client.create_database(db_name)
client.switch_database(db_name)

locations = ["1", "2", "3"]

while True:

    speed = {
        "measurement": "speed",
        "tags": {
            "location": random.choice(locations)
        },
        "fields": {
            "h": random.uniform(0, 100)
        }
    }

    density = {
        "measurement": "density",
        "tags": {
            "location": random.choice(locations)
        },
        "fields": {
            "g": random.uniform(0, 10)
        }
    }

    pressure = {
        "measurement": "pressure",
        "tags": {
            "location": random.choice(locations)
        },
        "fields": {
            "p": random.uniform(900, 1100)
        }
    }

    client.write_points([speed, density, pressure])

    print("Poslane 3 točke")

    time.sleep(0.5)
```
