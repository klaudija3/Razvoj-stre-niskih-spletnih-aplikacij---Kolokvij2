2. Naloga - InfluxDB branje
V Python napišite funkcijo ignoriraj tag(database, measurement, field, tag), ki
returna vse točke iz field od measurement iz database, ki nimajo taga tag.
Pričakovani format inputa funkcije tag naj bo slovar (npr. tag = {”asd” : ”dsa”}).

```python
from influxdb import InfluxDBClient


def ignoriraj_tag(database, measurement, field, tag):

    client = InfluxDBClient(
        host='149.62.71.186',
        port=8086,
        username='admin',
        password='fisinflux',
        database=database
    )

    tag_key = list(tag.keys())[0]
    tag_value = tag[tag_key]

    query = f'''
    SELECT "{field}"
    FROM "{measurement}"
    WHERE "{tag_key}" != '{tag_value}'
    '''

    result = client.query(query)

    return list(result.get_points())
```

```python
rezultat = ignoriraj_tag(
    "moje_ime_db",
    "speed",
    "h",
    {"location": "1"}
)

print(rezultat)
```
