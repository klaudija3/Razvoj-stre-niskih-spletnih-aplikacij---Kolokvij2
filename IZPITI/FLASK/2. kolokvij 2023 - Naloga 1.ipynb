1. Naloga - Flask
Z uporabo flask napišite program za pisanje in branje datoteke na strežniku.
1. Na naslovu /write/<filename> napišite POST metodo, s katero uporabnik pošlje
tekst. Strežnik naj tekst shrani v novo .txt datoteko, katere ime razbere iz naslova
(preko filename).
2. Na naslovu /read/< filename > napišite GET metodo, s katero uporabnik zahteva
za tekst, ki ga ima strežnik v .txt datoteki. Strežnik naj prebere datoteko, katere
ime razbere iz naslova (preko filename) in vsebino pošlje. Ce datoteka ne obstaja,
naj to sporoči.

```python
from flask import Flask, request

app = Flask(__name__)

# -------------------------
# WRITE
# -------------------------
@app.route('/write/<filename>', methods=['POST'])
def write_file(filename):

    try:
        text = request.data.decode()

        with open(f'{filename}.txt', 'w', encoding='utf-8') as f:
            f.write(text)

        return f'Datoteka {filename}.txt uspešno shranjena.'

    except Exception as e:
        return f'Napaka pri pisanju: {e}'


# -------------------------
# READ
# -------------------------
@app.route('/read/<filename>', methods=['GET'])
def read_file(filename):

    try:
        with open(f'{filename}.txt', 'r', encoding='utf-8') as f:
            data = f.read()

        return data

    except FileNotFoundError:
        return 'Datoteka ne obstaja.'


app.run(
    host='127.0.0.1',
    port=1234,
    debug=True,
    use_reloader=False
)
```

```python
import requests

res = requests.post(
    'http://127.0.0.1:1234/write/test',
    data='To je test'
)

print(res.text)
```

```python
import requests

res = requests.get(
    'http://127.0.0.1:1234/read/test'
)

print(res.text)
```
