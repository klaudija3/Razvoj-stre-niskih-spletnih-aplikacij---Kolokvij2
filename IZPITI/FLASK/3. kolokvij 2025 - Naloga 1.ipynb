Naloga 1: Spomnite se 2. naloge pri flask-vajah:
S flask napišite program za login. Program naj preveri, da je kombinacija username :
password pravilna. Imate torej tri funkcije: funkcija na /login ima template login.html.
Preko obrazca prejmemo username in password. Funkcija naj preveri, ali sta pravilna. Če
sta pravilna, naj uporabnika preusmeri na /success/username, sicer naj ga preusmeri na
/failure
Omenjali smo, da ima tako preprosta koda očitno ranljivost. Predeljate kodo tako, da se
izognite tej ranljivosti! Vse skupaj mora funkcionirati (preverite lokalno preko brskalnika).

```python
from flask import Flask
from flask import request
from flask import render_template
from flask import redirect
from flask import url_for

from werkzeug.security import generate_password_hash
from werkzeug.security import check_password_hash

app = Flask(__name__)

# -------------------------
# uporabniki
# -------------------------

users = {
    "martin": generate_password_hash("martin00"),
    "student": generate_password_hash("student00")
}

# -------------------------
# SUCCESS
# -------------------------

@app.route('/success/<name>')
def success(name):

    return render_template(
        'success.html',
        username=name
    )

# -------------------------
# FAILURE
# -------------------------

@app.route('/failure')
def failure():

    return render_template('failure.html')

# -------------------------
# LOGIN
# -------------------------

@app.route('/login', methods=['GET', 'POST'])
def login():

    if request.method == 'GET':
        return render_template('login.html')

    if request.method == 'POST':

        user = request.form['username']
        password = request.form['password']

        if user in users:

            if check_password_hash(users[user], password):

                return redirect(
                    url_for('success', name=user)
                )

        return redirect(
            url_for('failure')
        )

# -------------------------

app.run(
    host='127.0.0.1',
    port=1235,
    debug=True,
    use_reloader=False
)
```
