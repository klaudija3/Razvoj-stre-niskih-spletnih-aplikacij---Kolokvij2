3. Naloga - ASN.1
(a) Z uporabo knjižnice pyasn1 napišite class Oseba() za ASN.1 format, ki je oblike:
• String ”Ime”
• String ”Priimek”
• Integer ”Starost”, ki mora biti med 0 in 99
• Telefonsko številko, ki naj bo zaporedje devetih števil od 0 do 9 (kateri tip je za to
smiselno uporabiti, presodite sami).
(b) Naredite en primer tega formata, ga zakodirajte in pravilno dekodirajte.

a)

```python
from pyasn1.type import univ, namedtype, constraint, char
from pyasn1.codec.ber import encoder, decoder

# starost 0-99
class Starost(univ.Integer):
    subtypeSpec = univ.Integer.subtypeSpec + \
        constraint.ValueRangeConstraint(0, 99)

# telefon = natanko 9 številk
class Telefon(char.NumericString):
    subtypeSpec = char.NumericString.subtypeSpec + \
        constraint.ValueSizeConstraint(9, 9)

class Oseba(univ.Sequence):
    componentType = namedtype.NamedTypes(
        namedtype.NamedType('Ime', char.UTF8String()),
        namedtype.NamedType('Priimek', char.UTF8String()),
        namedtype.NamedType('Starost', Starost()),
        namedtype.NamedType('Telefon', Telefon())
    )
```

b)

```python
from pyasn1.codec.ber import encoder, decoder

oseba = Oseba()

oseba['Ime'] = 'Martin'
oseba['Priimek'] = 'Bresar'
oseba['Starost'] = 25
oseba['Telefon'] = '031123456'

print("Original:")
print(oseba)

# kodiranje
encoded = encoder.encode(oseba)

print("\nEncoded bytes:")
print(encoded)

# dekodiranje
decoded, rest = decoder.decode(encoded, asn1Spec=Oseba())

print("\nDecoded:")
print(decoded)
```
