```python
from influxdb import InfluxDBClient

host = '149.62.71.186'
user = 'admin'
password = 'fisinflux'
port = 8086

client = InfluxDBClient(host, port, user, password)

# unikaten database
db_name = "moje_ime_db"

client.create_database(db_name)
client.switch_database(db_name)

with open("temperature data.txt", "r") as f:

    for line in f:

        line = line.strip()

        # temperatura1,temperatura2,cas,oseba
        t1, t2, cas, oseba = line.split(",")

        point = {
            "measurement": "temperature",

            "tags": {
                "person": oseba
            },

            "time": cas,

            "fields": {
                "temperature1": float(t1),
                "temperature2": float(t2)
            }
        }

        client.write_points([point])

print("Podatki naloženi.")
```

```python
query = """
SELECT temperature2
FROM temperature
WHERE person='Eva'
AND temperature2 < 0
"""

result = client.query(query)

for point in result.get_points():
    print(point)
```
